## Recursion with Harry

```python
def show(n):
    if(n == 0):
        return
    print(n)
    show(n-1) 

show(5)
```

```python
def calc_sum(n):
    if (n == 0):
        return 0
    return calc_sum(n-1) + n
sum = calc_sum(10)
print(sum)
```

```python
def print_list(list, idx=0):
    if(idx == len(list)):
        return
    print_list(list, idx+1)
fruits = ["mango", "apple", "litch", "banana"]

print_list(fruits)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
n = int(input("Enter a number: "))
print("Factorial of", n, "is:", factorial(n))
```

```python
def power_sum(num, power):
    # Base case: when num becomes 0
    if num == 0:
        return 0
    else:
        # Extract last digit and add its power
        digit = num % 10
        return (digit ** power) + power_sum(num // 10, power)
 
def is_armstrong(num):
    # Count digits
    digits = len(str(num))
    # Compare original number with recursive sum
    return num == power_sum(num, digits)

# Taking input from user
n = int(input("Enter a number: "))

if is_armstrong(n):
    print(n, "is an Armstrong number.")
else:
    print(n, "is not an Armstrong number.")
```
