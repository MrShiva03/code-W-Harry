```python
import string

text = "Hello!!!, Shiva. How are you??"

result = text.translate(str.maketrans('', '', string.punctuation))

print(result)
```

```python
s = "shiva"

v = sum(1 for i in s if i in "aeiou")
c = sum(1 for i in s if i.isalpha() and i not in "aeiou")

print(v)
print(c)
```
